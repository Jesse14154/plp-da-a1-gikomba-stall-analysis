## Introduction

This notebook analyses 14 days of sales and stock records for Wanjiku's fresh-produce stall at Gikomba Market. The analysis uses core Python to examine sales performance, compare weekly results, classify daily performance, identify products requiring restocking, calculate gross margins, and provide business recommendations.

```python

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Calculate Sales Performance

In this task, I use Python's built-in functions to calculate total, average, highest, and lowest sales over the 14-day period. I also use list indexing to identify the day names and weeks in which the highest and lowest sales occurred.

```python

# Calculate total, average, highest and lowest sales

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Find the corresponding day names
best_day = days[best_index]
worst_day = days[worst_index]

# Calculate the week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

# Display the results
print("TASK 1: SALES PERFORMANCE")
print("-------------------------")
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {highest_sales:,.2f}")
print(f"Best day: {best_day}, Week {best_week}")
print(f"Lowest daily sales: KES {lowest_sales:,.2f}")
print(f"Worst day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales Comparison and Growth

In this task, I define a function called week_totals() to calculate the total sales for each week. I use list slicing to separate the 14 days into two weeks and calculate the percentage growth from Week 1 to Week 2.

```python

# Function to calculate weekly sales totals

def week_totals(sales):
    """
    Calculate the total sales for Week 1 and Week 2.

    Parameters:
        sales: A list containing 14 daily sales amounts.

    Returns:
        A tuple containing the Week 1 and Week 2 sales totals.
    """

    week1 = sum(sales[:7])
    week2 = sum(sales[7:])

    return week1, week2


# Call the function and store its returned values
week1_total, week2_total = week_totals(daily_sales)

# Calculate percentage growth
growth_percentage = ((week2_total - week1_total) / week1_total) * 100

# Display the results
print("TASK 2: WEEKLY SALES COMPARISON")
print("--------------------------------")
print(f"Week 1 total sales: KES {week1_total:,.2f}")
print(f"Week 2 total sales: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3: Classify Daily Sales Performance

In this task, I create a function called classify_day() to classify each day's sales as Strong, Normal, or Weak based on the average daily sales. I then use a dictionary to count the number of days in each category.

```python

# Function to classify daily sales performance

def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal, or Weak.

    Parameters:
        amount: The sales amount for a particular day.
        average: The average daily sales amount.

    Returns:
        Strong, Normal, or Weak depending on the sales amount.
    """

    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"


# Create a dictionary to count each category
performance_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify every day's sales and count the categories
daily_performance = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_performance.append(category)
    performance_counts[category] += 1


# Display each day's classification
print("TASK 3: DAILY SALES CLASSIFICATION")
print("-----------------------------------")

for i in range(len(daily_sales)):
    print(f"Day {i + 1} ({days[i]}): "
          f"KES {daily_sales[i]:,.2f} - {daily_performance[i]}")


# Display category totals
print("\nPERFORMANCE SUMMARY")
print("-------------------")

for category, count in performance_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday and Weekend Sales Comparison

In this task, I separate weekday sales from weekend sales using the day names. I calculate the average sales for each group and determine the percentage difference to establish whether the stall performs better during weekdays or weekends.

```python

# Separate weekday sales from weekend sales

weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):

    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])

    else:
        weekend_sales.append(daily_sales[i])


# Calculate average sales for each group

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)


# Calculate percentage difference

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100


# Display results

print("TASK 4: WEEKDAY AND WEEKEND COMPARISON")
print("---------------------------------------")
print(f"Weekday average sales: KES {weekday_average:,.2f}")
print(f"Weekend average sales: KES {weekend_average:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Conclusion: Weekend sales are higher than weekday sales.")
elif weekend_average < weekday_average:
    print("Conclusion: Weekday sales are higher than weekend sales.")
else:
    print("Conclusion: Weekday and weekend sales are equal.")
```

## Task 5: Stock Reordering and Gross Margin Analysis

In this task, I use the stock dictionary to identify products whose quantities are at or below their reorder levels. I also calculate the gross margin percentage for each product using its cost price and selling price.

```python

# Function to identify products that need restocking

def reorder_products(items):
    """
    Identify products whose stock quantity is at or below
    their reorder level.

    Parameters:
        items: A dictionary containing product stock information.

    Returns:
        A list of product names that need restocking.
    """

    products_to_reorder = []

    for product, details in items.items():

        if details["qty"] <= details["reorder"]:
            products_to_reorder.append(product)

    return products_to_reorder


# Call the function
to_reorder = reorder_products(stock)

# Display products requiring restocking
print("PRODUCTS REQUIRING RESTOCKING")
print("-----------------------------")

for product in to_reorder:
    print(product)
```

```python

# Calculate gross margin percentage for each product

print("GROSS MARGIN ANALYSIS")
print("---------------------")

for product, details in stock.items():

    cost_price = details["cost"]
    selling_price = details["price"]

    gross_margin = (
        (selling_price - cost_price) / selling_price
    ) * 100

    print(f"{product}: {gross_margin:.2f}%")
```

## Task 6: Stall Performance Report

In this task, I combine the results from the previous tasks into a clearly formatted Stall Performance Report. I use Python f-strings to display sales figures, weekly comparisons, performance classifications, stock alerts, and gross margins.

```python

# Generate the Stall Performance Report

print("=" * 55)
print("          STALL PERFORMANCE REPORT")
print("          Gikomba Market - Wanjiku")
print("=" * 55)

# 1. Overall sales performance
print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 55)

print(f"Total sales:          KES {total_sales:,.2f}")
print(f"Average daily sales:  KES {average_sales:,.2f}")
print(f"Highest sales:        KES {highest_sales:,.2f}")
print(f"Best sales day:       {best_day}, Week {best_week}")
print(f"Lowest sales:         KES {lowest_sales:,.2f}")
print(f"Worst sales day:      {worst_day}, Week {worst_week}")


# 2. Weekly comparison
print("\n2. WEEKLY SALES COMPARISON")
print("-" * 55)

print(f"Week 1 sales:         KES {week1_total:,.2f}")
print(f"Week 2 sales:         KES {week2_total:,.2f}")
print(f"Weekly growth:        {growth_percentage:.2f}%")


# 3. Daily performance classification
print("\n3. DAILY PERFORMANCE SUMMARY")
print("-" * 55)

for category, count in performance_counts.items():
    print(f"{category}:              {count} days")


# 4. Weekday and weekend comparison
print("\n4. WEEKDAY VS WEEKEND SALES")
print("-" * 55)

print(f"Weekday average:      KES {weekday_average:,.2f}")
print(f"Weekend average:      KES {weekend_average:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")


# 5. Stock reordering report
print("\n5. STOCK REORDERING ALERTS")
print("-" * 55)

if to_reorder:
    for product in to_reorder:
        print(f"REORDER: {product}")
else:
    print("No products currently require restocking.")


# 6. Gross margin report
print("\n6. PRODUCT GROSS MARGINS")
print("-" * 55)

for product, details in stock.items():

    gross_margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"{product}: {gross_margin:.2f}%")


# End of report
print("\n" + "=" * 55)
print("             END OF REPORT")
print("=" * 55)
```
